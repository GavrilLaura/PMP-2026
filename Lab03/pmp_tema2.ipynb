**Ex. 1**

Se aruncă două zaruri, iar rezultatele sunt notate cu X şi Y.
1. Calculaţi probabilitatea ca suma să fie cel puţin 10, ştiind că primul zar indică un număr par.
2. Calculaţi probabilitatea ca suma să fie cel puţin 10, ştiind că cel puţin unul dintre zaruri indică un număr par.
3. Calculaţi probabilitatea ca ambele zaruri să indice numere pare, ştiind că suma lor este 8.
4. Sunt evenimentele A = {X + Y ≥ 10} şi B = {X este par} independente?

Extensie Python: Simulaţi 100.000 de aruncări şi estimaţi probabilităţile condiţionate. Comparaţi rezultatele cu valorile exacte.

```python
# 1.1  P(X+Y >= 10 | X par)
import random

random.seed(1)
N = 100000

nr_x_par = 0        # aruncari in care X e par (conditia)
nr_favorabile = 0   # dintre acestea, cele cu suma >= 10

for i in range(N):
    x = random.randint(1, 6)
    y = random.randint(1, 6)

    if x % 2 == 0:
        nr_x_par = nr_x_par + 1
        if x + y >= 10:
            nr_favorabile = nr_favorabile + 1

print("Simulat:", round(nr_favorabile / nr_x_par, 4))
print("Exact:  ", round(2 / 9, 4))
```

```python
# 1.2  P(X+Y >= 10 | cel putin un zar par)
import random

random.seed(1)
N = 100000

nr_un_par = 0
nr_favorabile = 0

for i in range(N):
    x = random.randint(1, 6)
    y = random.randint(1, 6)

    if x % 2 == 0 or y % 2 == 0:
        nr_un_par = nr_un_par + 1
        if x + y >= 10:
            nr_favorabile = nr_favorabile + 1

print("Simulat:", round(nr_favorabile / nr_un_par, 4))
print("Exact:  ", round(5 / 27, 4))
```

```python
# 1.3  P(ambele pare | X+Y = 8)
import random

random.seed(1)
N = 100000

nr_suma_8 = 0       # aruncari cu suma 8 (conditia)
nr_favorabile = 0   # dintre acestea, cele cu ambele zaruri pare

for i in range(N):
    x = random.randint(1, 6)
    y = random.randint(1, 6)

    if x + y == 8:
        nr_suma_8 = nr_suma_8 + 1
        if x % 2 == 0 and y % 2 == 0:
            nr_favorabile = nr_favorabile + 1

print("Simulat:", round(nr_favorabile / nr_suma_8, 4))
print("Exact:  ", round(3 / 5, 4))
```

```python
# 1.4  Sunt A = {X+Y >= 10} si B = {X par} independente?
import random

random.seed(1)
N = 100000

nr_A = 0
nr_B = 0
nr_A_si_B = 0

for i in range(N):
    x = random.randint(1, 6)
    y = random.randint(1, 6)

    A = x + y >= 10
    B = x % 2 == 0

    if A:
        nr_A = nr_A + 1
    if B:
        nr_B = nr_B + 1
    if A and B:
        nr_A_si_B = nr_A_si_B + 1

p_A = nr_A / N
p_B = nr_B / N
p_A_si_B = nr_A_si_B / N

print("P(A) * P(B) simulat:", round(p_A * p_B, 4), "  exact:", round(1/12, 4))
print("P(A si B)   simulat:", round(p_A_si_B, 4), "  exact:", round(1/9, 4))
print("P(A | B)    simulat:", round(nr_A_si_B / nr_B, 4), "  vs  P(A) simulat:", round(p_A, 4))
```

**Ex. 2 *Filtrarea mesajelor spam***

Un sistem de e-mail primeşte 30% mesaje spam. Cuvântul „ofertă” apare în 80% dintre mesajele spam şi în 15% dintre mesajele legitime.
1. Care este probabilitatea ca un mesaj care conţine cuvântul „ofertă” să fie spam?
2. Presupunem că termenul „gratuit” apare în 60% dintre mesajele spam şi în 10% dintre mesajele legitime. Considerând independenţa condiţionată a apariţiei celor două cuvinte, calculaţi probabilitatea ca un mesaj care conţine ambele cuvinte să fie spam.
3. Explicaţi de ce ipoteza independenţei condiţionate poate fi nerealistă.
4. Propuneţi o regulă de clasificare bazată pe probabilitatea condiţionată.

```python
# 2.1  P(spam | "oferta")
p_spam = 0.3
p_oferta_daca_spam = 0.8
p_oferta_daca_legitim = 0.15

p_oferta = p_oferta_daca_spam * p_spam + p_oferta_daca_legitim * (1 - p_spam)
p_spam_daca_oferta = p_oferta_daca_spam * p_spam / p_oferta

print("P(oferta) =", round(p_oferta, 4))
print("P(spam | oferta) =", round(p_spam_daca_oferta, 4))
```

```python
# 2.2  P(spam | "oferta" si "gratuit"), cu independenta conditionata
p_spam = 0.3
p_oferta_daca_spam = 0.8
p_oferta_daca_legitim = 0.15
p_gratuit_daca_spam = 0.6
p_gratuit_daca_legitim = 0.1

# independenta conditionata: inmultim probabilitatile cuvintelor, pentru fiecare clasa
p_ambele_daca_spam = p_oferta_daca_spam * p_gratuit_daca_spam
p_ambele_daca_legitim = p_oferta_daca_legitim * p_gratuit_daca_legitim

p_ambele = p_ambele_daca_spam * p_spam + p_ambele_daca_legitim * (1 - p_spam)
p_spam_daca_ambele = p_ambele_daca_spam * p_spam / p_ambele

print("P(ambele cuvinte) =", round(p_ambele, 4))
print("P(spam | ambele cuvinte) =", round(p_spam_daca_ambele, 4))
```

### 2.3
Cuvintele dintr-un mesaj nu apar independent: în spam-ul promoțional apar des împreună („ofertă gratuită”). Deci, chiar știind că mesajul e spam, prezența lui „ofertă” crește șansa lui „gratuit”. Ipoteza ignoră această corelație, numără de două ori aceeași dovadă și dă probabilități prea extreme. Se folosește totuși pentru că simplifică mult calculul (Naive Bayes).

```python
# 2.4  Clasificam toate combinatiile de cuvinte (1 = apare, 0 = nu apare)
p_spam = 0.3
p_oferta_daca_spam = 0.8
p_oferta_daca_legitim = 0.15
p_gratuit_daca_spam = 0.6
p_gratuit_daca_legitim = 0.1
prag = 0.5

for oferta in [1, 0]:
    for gratuit in [1, 0]:
        # probabilitatea cuvintelor (prezenta sau absenta) pentru spam
        if oferta == 1:
            p_o_spam = p_oferta_daca_spam
        else:
            p_o_spam = 1 - p_oferta_daca_spam
        if gratuit == 1:
            p_g_spam = p_gratuit_daca_spam
        else:
            p_g_spam = 1 - p_gratuit_daca_spam

        # la fel pentru mesajele legitime
        if oferta == 1:
            p_o_leg = p_oferta_daca_legitim
        else:
            p_o_leg = 1 - p_oferta_daca_legitim
        if gratuit == 1:
            p_g_leg = p_gratuit_daca_legitim
        else:
            p_g_leg = 1 - p_gratuit_daca_legitim

        scor_spam = p_o_spam * p_g_spam * p_spam
        scor_legitim = p_o_leg * p_g_leg * (1 - p_spam)
        p_spam_cuvinte = scor_spam / (scor_spam + scor_legitim)

        if p_spam_cuvinte > prag:
            decizie = "SPAM"
        else:
            decizie = "legitim"

        print("oferta =", oferta, " gratuit =", gratuit,
              " P(spam) =", round(p_spam_cuvinte, 3), " ->", decizie)
```

**Ex. 3 *Paradoxul lui Simpson***

O universitate analizează rata de promovare la două discipline, A şi B, pentru studenţii din două programe de studiu:

**Programul 1:**

disciplina A -> 81 din 90
                 
disciplina B -> 19 din 20

**Programul 2:**

disciplina A -> 1 din 10

disciplina B -> 16 din 80

Valorile reprezintă numărul studenţilor promovaţi din totalul celor evaluaţi.
1. Calculaţi rata de promovare la fiecare disciplină, separat pentru fiecare program.
2. Care disciplină are o rată de promovare mai mare în fiecare program?
3. Calculaţi rata globală de promovare pentru cele două discipline.
4. Explicaţi de ce ordinea disciplinelor se inversează atunci când agregăm datele.
5. Formulaţi fenomenul în termeni de probabilităţi condiţionate.

Extensie Python: Construiţi un exemplu propriu de paradox Simpson şi utilizaţi librăria pandas pentru a analiza
datele.

```python
# 3.1  Rata de promovare pe discipline, separat pe programe
# (promovati, evaluati)
p1_A = (81, 90)
p1_B = (19, 20)
p2_A = (1, 10)
p2_B = (16, 80)

rata_p1_A = p1_A[0] / p1_A[1]
rata_p1_B = p1_B[0] / p1_B[1]
rata_p2_A = p2_A[0] / p2_A[1]
rata_p2_B = p2_B[0] / p2_B[1]

print("Programul 1:  A =", rata_p1_A, "  B =", rata_p1_B)
print("Programul 2:  A =", rata_p2_A, "  B =", rata_p2_B)
```

```python
# 3.2  Care disciplina are rata mai mare in fiecare program?
if rata_p1_B > rata_p1_A:
    print("Programul 1: B are rata mai mare")
else:
    print("Programul 1: A are rata mai mare")

if rata_p2_B > rata_p2_A:
    print("Programul 2: B are rata mai mare")
else:
    print("Programul 2: A are rata mai mare")
```

```python
# 3.3  Rata globala (adunam studentii din ambele programe)
promovati_A = p1_A[0] + p2_A[0]
evaluati_A = p1_A[1] + p2_A[1]
promovati_B = p1_B[0] + p2_B[0]
evaluati_B = p1_B[1] + p2_B[1]

rata_globala_A = promovati_A / evaluati_A
rata_globala_B = promovati_B / evaluati_B

print("Global A:", promovati_A, "/", evaluati_A, "=", rata_globala_A)
print("Global B:", promovati_B, "/", evaluati_B, "=", rata_globala_B)
```

```python
# 3.4  Rata globala ca medie ponderata a ratelor pe programe
pondere_p1_A = p1_A[1] / evaluati_A    # fractia studentilor de la A care sunt in Programul 1
pondere_p2_A = p2_A[1] / evaluati_A
pondere_p1_B = p1_B[1] / evaluati_B
pondere_p2_B = p2_B[1] / evaluati_B

print("A: ponderi P1 =", pondere_p1_A, ", P2 =", pondere_p2_A)
print("B: ponderi P1 =", pondere_p1_B, ", P2 =", pondere_p2_B)

medie_ponderata_A = rata_p1_A * pondere_p1_A + rata_p2_A * pondere_p2_A
medie_ponderata_B = rata_p1_B * pondere_p1_B + rata_p2_B * pondere_p2_B

print("A: medie ponderata =", round(medie_ponderata_A, 2), " (egal cu rata globala", rata_globala_A, ")")
print("B: medie ponderata =", round(medie_ponderata_B, 2), " (egal cu rata globala", rata_globala_B, ")")
```

### 3.5
Fie R = promovare, D = disciplina, G = programul.
Pe fiecare program: P(R | B, G=g) > P(R | A, G=g), pentru g = 1, 2.
Global: P(R | A) > P(R | B).

Din formula probabilității totale:
P(R | D) = P(R | D, G=1)·P(G=1 | D) + P(R | D, G=2)·P(G=2 | D)

Paradoxul apare deoarece ponderile P(G | D) diferă între discipline: P(G=1 | A) = 0.9, dar P(G=1 | B) = 0.2. Dacă G ar fi independent de D, inversarea nu ar putea avea loc.

```python
import pandas as pd

date = {
    "tratament": ["X", "X", "Y", "Y"],
    "grup":      ["usor", "grav", "usor", "grav"],
    "vindecati": [160, 10, 45, 75],
    "total":     [200, 50, 50, 200],
}

df = pd.DataFrame(date)
df["rata"] = df["vindecati"] / df["total"]
print(df)

# randuri = grupuri, coloane = tratamente, valori = rata de vindecare
comparatie = df.pivot(index="grup", columns="tratament", values="rata")
print(comparatie)

# adunam vindecatii si totalul pentru fiecare tratament, ignorand grupul
globala = df.groupby("tratament")[["vindecati", "total"]].sum()
globala["rata"] = globala["vindecati"] / globala["total"]
print(globala)
```
