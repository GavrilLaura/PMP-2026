##**Formula lui Bayes**

Să presupunem că investigăm o boală (B), care afectează doar un procent din populaţie (P(B) = 0.01). Avem un test de diagnosticare pentru această boală, care are următoarele caracteristici:

● Sensibilitate (probabilitatea ca testul să fie pozitiv dacă persoana este bolnavă): P(Test = Pozitiv∣B) = 0.95

● Specificitate (probabilitatea ca testul să fie negativ dacă persoana nu este bolnavă): P(Test = Negativ∣¬B) = 0.90


**a)** Dacă o persoană este testată şi rezultatul este pozitiv, care este probabilitatea ca aceasta să fie efectiv bolnavă?
Cum explicaţí rezultatul?

**b)** Modelaţi în Python probabilitatea de mai sus şi comparaţi cu cea obţinută teoretic la punctul anterior.

**c)** Care ar trebui să fie specificitatea minimă pentru ca probabilitatea de mai sus să ajungă la 50%?

```python
# a)
p_B = 0.01
sens = 0.95 # P(test_pozitiv|B)
spec = 0.90 # P(test_negativ|~B)

p_test_pozitiv = sens * p_B + (1 - spec) * (1 - p_B)    # P(A)
p_B_test_pozitiv_teoretic = sens * p_B / p_test_pozitiv
print(f"P(T+)     = {p_test_pozitiv:.4f}")
print(f"P(B | T+) teoretic = {p_B_test_pozitiv_teoretic:.4f}")
```

```python
# b)
import random

p_B = 0.01      # probabilitatea sa fii bolnav
sens = 0.95     # P(test pozitiv | bolnav)
spec = 0.90     # P(test negativ | sanatos)
N = 1000000     # numarul de persoane simulate

pozitive = 0            # nr total teste pozitive
bolnavi_pozitivi = 0    # nr teste pozitive la persoane bolnave

for i in range(N):
  if random.random() < p_B:
    bolnav = True
  else:
    bolnav = False

  if bolnav:
    test_pozitiv = random.random() < sens #test pozitiv adevarat cu prob 0.95
  else:
    test_pozitiv = random.random() < (1 - spec) #test pozitiv fals cu prob 0.10

  if test_pozitiv:
    pozitive = pozitive + 1
    if bolnav:
      bolnavi_pozitivi = bolnavi_pozitivi + 1

p_sim = bolnavi_pozitivi / pozitive
print("Simulat  P(B | T+) =", p_sim)
print("Teoretic P(B | T+) =", 0.0876)
```

## c) Specificitatea minimă

Notăm specificitatea cu s. Vrem P(B | T+) ≥ 0.5:

$$\frac{0.95 \cdot 0.01}{0.95 \cdot 0.01 + (1 - s)\cdot 0.99} \geq 0.5$$

Înmulțim cu numitorul (care este pozitiv):

$$0.0095 \geq 0.5 \cdot 0.0095 + 0.5 \cdot (1 - s) \cdot 0.99$$

$$0.00475 \geq 0.495\,(1 - s) \quad\Rightarrow\quad 1 - s \leq 0.0096 \quad\Rightarrow\quad s \geq 0.9904$$

**Specificitatea trebuie să fie de cel puțin ≈ 99.04%**, adică testul poate da alarme false la cel mult ~1% din oamenii sănătoși.

Intuitiv: P(B | T+) = 0.5 înseamnă că printre testele pozitive există tot atâția bolnavi adevărați câte alarme false.
