#**Exerciţiu laborator 1**
##**Simularea variabilelor aleatoare**

Considerăm o urnă în care avem 3 bile roşii, 4 albastre şi 2 negre. Aruncăm un zar; dacă obţinem un număr prim, adăugăm o bilă neagră în urnă, dacă obţinem 6, adăugăm o bilă roşie, iar în celelalte cazuri adăugăm o bilă albastră. Apoi extragem o bilă din urnă.

**a)** Simulaţi în Python experimentul de mai sus.

**b)** Folosind modelarea făcută, estimaţi probabilitatea de a obţine o bilă roşie.

**c)** Calculaţi probabilitatea teoretică a evenimentului de mai sus şi comparaţi-o cu cea obţinută.


```python
import random
import numpy as np
import matplotlib.pyplot as plt
from fractions import Fraction

random.seed(42)

URNA_INITIALA = {"rosu": 3, "albastru": 4, "negru": 2}
NUMERE_PRIME = {2, 3, 5}
```

```python
#a)

def experiment():
    urna = URNA_INITIALA.copy()

    zar = random.randint(1, 6)
    if zar in NUMERE_PRIME:
        urna["negru"] += 1
    elif zar == 6:
        urna["rosu"] += 1
    else:
        urna["albastru"] += 1

    culori = list(urna.keys())
    nr_bile = list(urna.values())
    bila = random.choices(culori, weights=nr_bile)[0]

    return zar, bila

for i in range(5):
    zar, bila = experiment()
    print(f"Experimentul {i+1}: zar = {zar}, bila extrasa = {bila}")
```

```python
#b)

N = 100_000
nr_rosii = 0

for _ in range(N):
    zar, bila = experiment()
    if bila == "rosu":
        nr_rosii += 1

p_estimat = nr_rosii / N
print(f"Din {N} experimente, bila rosie a aparut de {nr_rosii} ori.")
print(f"Probabilitatea estimata P(rosu) ≈ {p_estimat:.4f}")
```

## c) Calculul teoretic

Notăm cu $R$ evenimentul „extragem o bilă roșie” și împărțim după rezultatul zarului (formula probabilității totale):

| Rezultat zar | Probabilitate | Bila adăugată | Urna (R, A, N) | $P(R \mid \text{zar})$ |
|---|---|---|---|---|
| prim: 2, 3, 5 | $3/6 = 1/2$ | neagră | (3, 4, 3) | $3/10$ |
| 6 | $1/6$ | roșie | (4, 4, 2) | $4/10$ |
| 1 sau 4 | $2/6 = 1/3$ | albastră | (3, 5, 2) | $3/10$ |

$$P(R) = \frac{1}{2}\cdot\frac{3}{10} + \frac{1}{6}\cdot\frac{4}{10} + \frac{1}{3}\cdot\frac{3}{10} = \frac{9}{60} + \frac{4}{60} + \frac{6}{60} = \frac{19}{60} \approx 0.3167$$

```python
P_prim = Fraction(3, 6)
P_sase = Fraction(1, 6)
P_rest = Fraction(2, 6)

p_teoretic = (P_prim * Fraction(3, 10)
              + P_sase * Fraction(4, 10)
              + P_rest * Fraction(3, 10))

print(f"Probabilitatea teoretica: {p_teoretic} ≈ {float(p_teoretic):.4f}")
print(f"Probabilitatea estimata:  {p_estimat:.4f}")
print(f"Diferenta absoluta:       {abs(p_estimat - float(p_teoretic)):.4f}")
```
